## 1. Mount Google Drive

## 2. Clone / Initialise the GitHub Repository

````python
# Fill these in with your own values before running
GITHUB_USER  = 'your-username'
GITHUB_REPO  = 'ai-smart-retail'
GITHUB_TOKEN = 'ghp_xxx_replace_with_your_PAT'  # ⚠️ never commit this

import os, subprocess
os.makedirs(PROJECT_ROOT, exist_ok=True)
os.chdir(PROJECT_ROOT)

if not os.path.isdir(os.path.join(PROJECT_ROOT, '.git')):
    subprocess.run(['git', 'init'], check=True)
    subprocess.run(['git', 'config', 'user.email', 'you@example.com'], check=True)
    subprocess.run(['git', 'config', 'user.name',  GITHUB_USER], check=True)
    remote = f'https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{GITHUB_REPO}.git'
    subprocess.run(['git', 'remote', 'add', 'origin', remote])
print('Repository initialised at', PROJECT_ROOT)
````

## 3. Create the Standard Folder Structure

````python
FOLDERS = [
    'data/products', 'data/faces', 'data/reviews',
    'models', 'src', 'notebooks', 'tests',
]
for f in FOLDERS:
    os.makedirs(os.path.join(PROJECT_ROOT, f), exist_ok=True)

!ls -la {PROJECT_ROOT}
````
