## 1. Image Utilities

```python
from src.image_utils import load_image, resize_image, to_grayscale, normalize_image, augment_image
import matplotlib.pyplot as plt
import cv2

img = load_image('data/products/shirts/img_00.jpg')

fig, axes = plt.subplots(1, 4, figsize=(12, 3))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB));                 axes[0].set_title('Original')
axes[1].imshow(cv2.cvtColor(resize_image(img, (96, 96)), cv2.COLOR_BGR2RGB)); axes[1].set_title('Resized')
axes[2].imshow(to_grayscale(img), cmap='gray');                       axes[2].set_title('Gray')
axes[3].imshow(cv2.cvtColor(augment_image(img, flip=True, rotate_deg=12), cv2.COLOR_BGR2RGB)); axes[3].set_title('Augmented')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

## 2. Face Detection & Enrolment

```python
from src.face_utils import FaceRecognizer

fr = FaceRecognizer()
n = fr.enroll_from_folder('data/faces')
print(f'Enrolled {n} customers ->', list(fr.db.keys()))
```

## 3. Identification Test
