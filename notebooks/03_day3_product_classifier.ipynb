# Day 3 — Deep Learning Product Classifier

**Goal**: Train and evaluate a MobileNetV2 transfer-learning CNN that classifies
retail product images (shirts / shoes / bags in the sample dataset).

**Deliverable**: `03_day3_product_classifier.ipynb`


```python
import os, sys
PROJECT_ROOT = '/content/drive/MyDrive/Major_Project'
sys.path.append(PROJECT_ROOT); os.chdir(PROJECT_ROOT)
```

## 1. Inspect Dataset

```python
from pathlib import Path
for cls in sorted(os.listdir('data/products')):
    print(cls, '->', len(list(Path(f'data/products/{cls}').glob('*.jpg'))))
```

## 2. Train MobileNetV2 Classifier

```python
from src.product_classifier import train_model

model, history, labels = train_model(
    data_dir='data/products',
    model_out='models/product_classifier_model.h5',
    labels_out='models/product_labels.json',
    epochs=5,
    batch_size=8,
)
print('Classes:', labels)
```

## 3. Accuracy / Loss Curves

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(history.history['accuracy'], label='train')
axes[0].plot(history.history['val_accuracy'], label='val')
axes[0].set_title('Accuracy'); axes[0].legend()
axes[1].plot(history.history['loss'], label='train')
axes[1].plot(history.history['val_loss'], label='val')
axes[1].set_title('Loss'); axes[1].legend()
plt.show()
```

## 4. Inference on a Single Image

```python
from src.product_classifier import predict_image
label, conf = predict_image('data/products/shoes/img_00.jpg')
print(f'Predicted: {label}  (confidence={conf:.2%})')
```
