# Day 4 — NLP Sentiment Analysis & Customer Support Chatbot

**Goal**: Train a TF-IDF + Logistic-Regression sentiment classifier on retail reviews,
and build a retrieval-based FAQ chatbot for customer support.

**Deliverable**: `04_day4_sentiment_chatbot.ipynb`


```python
import os, sys
PROJECT_ROOT = '/content/drive/MyDrive/Major_Project'
sys.path.append(PROJECT_ROOT); os.chdir(PROJECT_ROOT)
```

## 1. Load Reviews

```python
import pandas as pd
df = pd.read_csv('data/reviews/reviews.csv')
print(df.shape); df.head()
```

## 2. Train the Sentiment Model

```python
from src.sentiment_model import SentimentModel

model = SentimentModel()
report = model.train(df)
print('Accuracy:', round(report['accuracy'], 3))
model.save('models/sentiment_model.joblib')
```

## 3. Try It Out

```python
for text in [
    'Absolutely love this product, best purchase ever!',
    'The bag strap broke on day two, very disappointed.',
    'It is okay, nothing special.',
]:
    label, prob = model.predict(text)
    print(f'{label:8s}  ({prob:.2f})  -> {text}')
```

## 4. Retail FAQ Chatbot

```python
from src.chatbot import RetailChatbot

bot = RetailChatbot()
bot.save('models/chatbot.joblib')

for q in [
    'How do I return a product I bought last week?',
    'What time do stores close on Sunday?',
    'Do you deliver internationally?',
    'zaqwsxcderfv',
]:
    ans, score = bot.reply(q)
    print(f'\nQ: {q}\nA ({score:.2f}): {ans}')
```
